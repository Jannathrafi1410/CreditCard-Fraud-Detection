## 1. Environment Setup & Library Imports

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization styling
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 6)
print('Libraries successfully imported!')
```

## 2. Ingesting the Dataset
We load the transaction dataset and examine its initial dimensions and column structure.

```python
data_path = '../data/raw/creditcard.csv'
if not os.path.exists(data_path):
    data_path = '../data/sample/sample_transactions.csv'

df = pd.read_csv(data_path)
print(f'Dataset Shape: {df.shape[0]:,} rows, {df.shape[1]} columns')
df.head()
```

## 3. Investigating Extreme Class Imbalance
Credit card fraud is rare. Detecting fraudulent transactions without overwhelming human fraud analysts with false alarms is the primary challenge.

```python
class_counts = df['Class'].value_counts()
fraud_pct = (class_counts[1] / len(df)) * 100
legit_pct = (class_counts[0] / len(df)) * 100

print(f'Legitimate Transactions (0): {class_counts[0]:,} ({legit_pct:.4f}%)')
print(f'Fraudulent Transactions (1): {class_counts[1]:,} ({fraud_pct:.4f}%)')
print(f'Imbalance Ratio: 1 : {class_counts[0] // class_counts[1]}')

plt.figure(figsize=(7, 4))
sns.barplot(x=class_counts.index, y=class_counts.values, hue=class_counts.index, palette=['#3b82f6', '#ef4444'], legend=False)
plt.yscale('log')
plt.title('Class Distribution (Log Scale)', fontsize=14, fontweight='bold')
plt.xlabel('Class (0: Legit, 1: Fraud)')
plt.ylabel('Count (Log10)')
plt.show()
```

## 4. Transaction Amount Profiling
We analyze how transaction amounts differ between legitimate and fraudulent activities.

```python
print('--- Amount Summary Statistics by Class ---')
display(df.groupby('Class')['Amount'].describe().T)

plt.figure(figsize=(8, 4))
sns.boxplot(x='Class', y='Amount', data=df, hue='Class', palette=['#93c5fd', '#fca5a5'], showfliers=False, legend=False)
plt.title('Transaction Amount by Class (Excluding Outliers)', fontsize=14, fontweight='bold')
plt.xticks([0, 1], ['Legitimate (0)', 'Fraudulent (1)'])
plt.show()
```

## 5. Temporal Patterns Across 48 Hours
The `Time` column records elapsed seconds. We examine whether fraud displays temporal periodicity.

```python
plt.figure(figsize=(10, 4))
sns.kdeplot(df[df['Class'] == 0]['Time'], label='Legitimate (0)', fill=True, color='#2563eb', alpha=0.3)
sns.kdeplot(df[df['Class'] == 1]['Time'], label='Fraudulent (1)', fill=True, color='#dc2626', alpha=0.4)
plt.title('Transaction Temporal Density Across 48 Elapsed Hours', fontsize=14, fontweight='bold')
plt.xlabel('Time (Elapsed Seconds)')
plt.legend()
plt.show()
```

## 6. Feature Correlation with Target Class
We examine the strongest linear correlations with fraud.

```python
corr = df.corr()
top_corr = corr['Class'].abs().sort_values(ascending=False).head(15)
print('Top Correlated Features with Class:')
print(top_corr)

plt.figure(figsize=(10, 8))
sns.heatmap(df[top_corr.index].corr(), cmap='coolwarm', annot=True, fmt='.2f', vmin=-1, vmax=1)
plt.title('Top 15 Features Correlation Heatmap', fontsize=14, fontweight='bold')
plt.show()
```
